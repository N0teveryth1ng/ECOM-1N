```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(r"C:\Users\S Das\Downloads\archive (2)\Amazon Sale Report.csv", low_memory=False)

# Clean and drop unnecessary unnamed columns
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]

# Create necessary calculated columns
df['Total Revenue (by Qty)'] = df['Qty'] * df['Amount']
df['Total Revenue (by Order)'] = df.groupby('Order ID')['Amount'].transform('sum')
df['Profit'] = df['Recived Amount'] - df['Expance']

# Optional: Add summary MRP difference if relevant
df['Amazon vs Flipkart MRP'] = df['Amazon MRP'] - df['Flipkart MRP']
df['Amazon vs Myntra MRP'] = df['Amazon MRP'] - df['Myntra MRP']

# Optional: Add any logic to clean or filter further (e.g., dropping NaNs if needed)
df = df.dropna(subset=['Qty', 'Amount'])

# Save everything to a single CSV file
output_path = r"C:\Users\S Das\Documents\PowerBI\Final_Export.csv"
df.to_csv(output_path, index=False, encoding='utf-8')

print("✅ Exported to:", output_path)
```
